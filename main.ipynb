```python
# Data manipulation
import pandas as pd
import numpy as np

# Data visualization
import matplotlib.pyplot as plt

# Show all columns
pd.set_option("display.max_columns", None)
```

```python
flipkart = pd.read_excel("Flipkart Sales Dataset(raw data).xlsx")
```

```python
# Dataset inspection
print("First 5 rows:", flipkart.head())

# Last 5 rows
print("Last 5 rows:", flipkart.tail())

# Dataset size
print("\n Dataset size:", flipkart.shape)

# Column Names
print("\n Data column:", flipkart.columns)

# Data types
flipkart.info()
```

```python
# Check missing values 
print("Missing values:", flipkart.isnull().sum())
```

```python
# Calculate missing values percentage
missing = flipkart.isnull().sum() / len(flipkart) * 100
missing = missing.sort_values(ascending=False)
print(missing)
```

```python
# Check duplicate rows 
flipkart.duplicated().sum()
```

```python
# Data types
flipkart.info()
```

```python
# Basic sattistics
flipkart.describe()
```

```python
# Total sales
total_sales = flipkart["Sale Price"].sum()

print("Total Sales:", total_sales)
```

```python
# Total orders
total_orders = flipkart["OrderID"].nunique()

print("Total Orders:", total_orders)
```

```python
# Average order value
average_order_values = flipkart["Sale Price"].mean()

print("Average Order Value:", round(average_order_values, 2))
```

```python
# Total_quantity sold
total_quqntity = flipkart["Order Quantity"].sum()

print("Total Quantity Sold:", total_quqntity)
```

```python
# Category-wise sales
category_sales = flipkart.groupby("Product Category")["Sale Price"].sum()

print(category_sales)
```

```python
# Sort highest to lowest
category_sales = category_sales.sort_values(ascending = False)

print(category_sales)
```

```python
# Plot a bar chart
top_10 = category_sales.head()

plt.Figure(figsize = (10,6))
plt.bar(top_10.index, top_10.values)

plt.title("Top 10 Product Categories by Sales")
plt.xlabel("Product Category")
plt.ylabel("Total Sales")

plt.xticks(rotation = 45)

plt.tight_layout()
plt.show()
```

```python
product_sales = flipkart.groupby("Product")["Sale Price"].sum()

product_sales = product_sales.sort_values(ascending = False)

print(product_sales.head(10))
```

```python
# Create bar chart
top_10_products = product_sales.head(10)

plt.Figure(figsize = (12,6))

plt.bar(top_10_products.index,top_10_products.values)

plt.title("Top 10 Products by Sales")
plt.xlabel("Product")
plt.ylabel("Total Sales")

plt.xticks(rotation = 90)

plt.tight_layout()
plt.show()
```

```python
# Extract month
flipkart["Month"] = flipkart["OrderDate"].dt.month_name()

# Calculate Monthly sales
monthly_sales = flipkart.groupby("Month")["Sale Price"].sum()

print(monthly_sales)
```

```python
month_order = [
    "January", "February", "March", "April",
    "May", "June", "July","August",
    "September", "Ortober", "November",
    "December"
]

monthly_sales = monthly_sales.reindex(month_order)
```

```python
# Create line chart
plt.figure(figsize = (10,6))

plt.plot(monthly_sales.index,
         monthly_sales.values,
         marker = "o",
         linewidth = 2)

plt.title("Monthly Sales Trend")
plt.xlabel("Month")
plt.ylabel("Total Sales")

plt.xticks(rotation = 45)
plt.grid(True)
plt.tight_layout()
plt.show()
```

```python
# Calculate location-wise Sales
location_sales = flipkart.groupby("Location")["Sale Price"].sum()

location_sales = location_sales.sort_values(ascending = False)

print(location_sales.head(10))
```

```python
# Create a Horizontal bar chart
top_10_locations = location_sales.head(10)

plt.figure(figsize = (10,6))

plt.barh(top_10_locations.index, top_10_locations.values)

plt.title("Top 10 Location by Sales")
plt.xlabel("Total Sales")
plt.ylabel("Location")

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
# Check age distribution
flipkart["Customer Age"].describe()
```

```python
# Create age groups
bins = [0, 18, 25, 35, 45, 60, 100]

labels = [
    "Under 18",
    "18-25",
    "26-35",
    "36-45",
    "46-60",
    "60+"
]

flipkart["Age Group"] = pd.cut(flipkart["Customer Age"],
    bins=bins,
    labels=labels
)
```

```python
# Count customers in each age group
age_group = flipkart["Age Group"].value_counts().sort_index()

print(age_group)
```

```python
# Create bar chart
plt.Figure(figsize = (8,5))

plt.bar(age_group.index, age_group.values)

plt.title("Customers by Age Group")
plt.xlabel("Age Group")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()
```

```python
# Delivery Type-Wise Sales
delivery_sales = flipkart.groupby("Delivery Type")["Sale Price"].sum()

delivery_sales = delivery_sales.sort_values(ascending = False)

print(delivery_sales)
```

```python
# Create bar chart
plt.Figure(figsize = (8,5))

plt.bar(delivery_sales.index, delivery_sales.values)

plt.title("Sales by Delivery Type")
plt.xlabel("Delivery Type")
plt.ylabel("Total Sales")

plt.xticks(rotation = 15)
plt.tight_layout()
plt.show()
```

```python
# Average rating by Delivery type
delivery_rating = flipkart.groupby("Delivery Type")["Rating"].mean()

print(delivery_rating)
```

```python
# Rating chart
plt.figure(figsize = (8,5))

plt.bar(delivery_rating.index, delivery_rating.values)

plt.title("Average Customer Rating by Delivery Type")
plt.xlabel("Delivery Type")
plt.ylabel("Average Rating")

plt.xticks(rotation = 15)
plt.tight_layout()
plt.show()
```

```python
# Count orders by status
status_count = flipkart["Status"].value_counts()

print(status_count)
```

```python
# Create bar chart
plt.Figure(figsize = (7,5))

plt.bar(status_count.index, status_count.values)

plt.title("Order Status Distribution")
plt.xlabel("Order Status")
plt.ylabel("Number of Orders")

plt.tight_layout()
plt.show()
```

```python
# Calculate returns rate
return_rate = ((flipkart["Status"] == "Returned").sum() / len(flipkart)) * 100

print(f"return Rate: {return_rate:.2f}%")
```

```python
# Calculate Cancellation rate
cancellation_rate = ((flipkart["Status"] == "Cancelled").sum() / len(flipkart)) * 100

print(f"Cancellation Rate: {cancellation_rate:.2f}%")
```

```python
# Analyze Return/cancel Reasons
reason_count = flipkart["Reason"].value_counts()

print(reason_count.head())
```

```python
# Gross Sales
gross_sales = flipkart["Sale Price"].sum()

print(f"Gross Sales: ₹{gross_sales:,.2f}")
```

```python
# Net Sales
delivered_orders = flipkart[flipkart["Status"] == "Delivered"]

net_sales = delivered_orders["Sale Price"].sum()

print(f"Net Sales: ₹{net_sales:,.2f}")
```

```python
# Revenue Lost
revenue_loss = gross_sales - net_sales

print(f"Revenue Loss: ₹{revenue_loss:,.2f}")
```

```python
# Revenue by Order Status
status_sales = flipkart.groupby("Status")["Sale Price"].sum()

print(status_sales)
```

```python
# Visualization
plt.Figure(figsize = (8,5))

plt.bar(status_sales.index, status_sales.values)

plt.title("Revenue by Order Status")
plt.xlabel("Order Status")
plt.ylabel("Revenue")

plt.tight_layout()
plt.show()
```

```python
# Select Numeric Columns
numeric_data = flipkart.select_dtypes(include = ["int64", "float64"])

numeric_data.head()
```

```python
# Calculate Correlation Matrix
correlation = numeric_data.corr()

print(correlation)
```

```python
# importing seaborn
import seaborn as sns

# visualize with heatmap
plt.Figure(figsize = (10,8))

sns.heatmap(
    correlation,
    annot = True,
    cmap = "coolwarm",
    fmt = ".2f"
)

plt.title("Correlation Matrix")
plt.show()
```

```python
# Total Unique Customer
total_customers = flipkart["CustomerID"].nunique()

print("Total Unique Customers:", total_customers)
```

```python
# Top 10 Customer by Revenue
top_customers = (flipkart.groupby("CustomerID")["Sale Price"]
                 .sum()
                 .sort_values(ascending = False)
                 .head(10)
)

print(top_customers)
```

```python
# Horizontal Bar Chart
plt.Figure(figsize = (10,6))

plt.barh(
    top_customers.index.astype(str),
    top_customers.values
)

plt.title("Top 10 Customers by Revenue")
plt.xlabel("Revenue")
plt.ylabel("Customer ID")

plt.gca().invert_yaxis()
plt.tight_layout()
plt.show()
```

```python
# Repeat vs one - time Customers
customer_orders = flipkart.groupby("CustomerID")["OrderID"].count()

repeat_customers = (customer_orders > 1).sum()
one_time_customers = (customer_orders == 1).sum()

print("Repeat Customers:", repeat_customers)
print("One-Time Customers:", one_time_customers)
```
